### Business Scenario
Data is the fuel for decision making. Mysql is a widely used relational database management system that helps businesses to store and manage large amounts of data.

```python
import getpass
import mysql.connector as mysql
db_password=getpass.getpass("password")
db=mysql.connect(
    host='localhost',
    user='root',
    password=db_password,
    port=3307
)
cursor=db.cursor()
query="CREATE DATABASE COMMERCE"
cursor.execute(query)
query0="USE COMMERCE"
cursor.execute(query0)
# create table customers
customers="CREATE TABLE CUSTOMERS(customer_id INT PRIMARY KEY,firstname VARCHAR(100),lastname VARCHAR(100),email VARCHAR(100),phone VARCHAR(100))"
cursor.execute(customers)
test="DESC CUSTOMERS"
cursor.execute(test)
print()
# create table orders
orders="CREATE TABLE ORDERS(order_id INT PRIMARY KEY,customer_id INT,orderdate DATE,totalamount DECIMAL(10,2),FOREIGN KEY (customer_id) REFERENCES CUSTOMERS(customer_id))"
cursor.execute(orders)
test1="DESC ORDERS"
cursor.execute(test1)
print()
#create inventory table
inventory="CREATE TABLE INVENTORY(product_id INT PRIMARY KEY,productname VARCHAR(100),quantity INT,price DECIMAL(10,2))"
cursor.execute(inventory)
test2="DESC INVENTORY"
cursor.execute(test2)
cursor.close()
db.close()
```
