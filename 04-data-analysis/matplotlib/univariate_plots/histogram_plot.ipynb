```python
import numpy as np
import matplotlib.pyplot as plt
from random import  randint
data=[]
for i in range (100):
    value=randint(0,100)
    data.append(value)
print(data)
print()
plt.hist(data,bins='auto',color='green',alpha=0.7)
plt.xlabel('value')
plt.ylabel('frequency')
plt.title('plotting random integers')
plt.show()
```
