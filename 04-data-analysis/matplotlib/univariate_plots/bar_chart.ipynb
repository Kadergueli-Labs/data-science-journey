### Bar chart
A bar chart or graph that presents categorical data with rectangular bars.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from pandas import Series,DataFrame
data=[20,40,17,23,50,89,80,90,32,45,10]
x=range(1,12)
plt.bar(x,data)
plt.show()
```

### Problem statement 3
Write a program that draws horizontal bar graph depicting marks obtained by a studentt in different subjects.
The barh() function is used to plot a horizontal bar gr

```python
import numpy as np
import matplotlib.pyplot as plt

subjects=["mathematics","computers","english","accountancy","economic"]
performance=[78,80,100,90,65]
y_pos=np.arange(len(subjects))
plt.barh(y_pos,performance,align='center')
plt.yticks(y_pos,subjects)
plt.title("mark obtained in different subjects")
plt.xlabel("marks")
plt.show()
```
