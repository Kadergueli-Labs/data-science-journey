### Bar chart
A bar chart or graph that presents categorical data with rectangular bars.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from pandas import Series,DataFrame
data=[20,40,17,23,50,89,80,90,32,45,10]
x=range(1,12)
plt.bar(x,data)
plt.show()
```
