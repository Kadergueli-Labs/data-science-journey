### Pandas
Pandas is an open source Python library that functions provideing high performance data manipulation and analysis.

### creating a data frame
A data frame is a two dimensional data structure im which data are organize in a tabular fashion using rows and columns.
syntax: pandas.DataFrame(data,index,columns,dtype,copy)

```python
#creating empty dataframe
import pandas as pd
import numpy as np
data=pd.DataFrame()
print(data)
```

```python
# creating dataframe from list
import pandas as np
import numpy as np
a=[1,2,3,4,5,6]
data=pd.DataFrame(a)
print(data)
```

```python
# creating dataframe from a list of list
import numpy as np
import pandas as pd
a=[1,["a",3],4,["b",5],6]
data=pd.DataFrame(a)
print(data)
```

### creation daframe from dictionary
Write a program that creates a dataframe storing names and marks of students using list of lists. give relevant names to the columns

```python
import numpy as np
import pandas as pd
a=[["Emie",18],["Marie",17],["Joel",20],["Lorielle",20],["Joseph",12]]
data=pd.DataFrame(a,columns=["Name","Marks"])
print(data)
```

Write a program that creates a data frame from a dictionary of ndarrays and lists

Write a program that creates a dataframe from a list of dictionary

```python
import pandas as pd
import numpy as np
data=[{"Name":'Marie',"Marks":12},{"Name":'Joseph',"Mark":80},{"Name":'Ali',"marks":70}]
df=pd.DataFrame(data,index=["student_1","student_2","student_3"])
print(df)
```
