### Pandas
Pandas is an open source Python library that functions provideing high performance data manipulation and analysis.

### creating a data frame
A data frame is a two dimensional data structure im which data are organize in a tabular fashion using rows and columns.
syntax: pandas.DataFrame(data,index,columns,dtype,copy)

```python
#creating empty dataframe
import pandas as pd
import numpy as np
data=pd.DataFrame()
print(data)
```

```python
# creating dataframe from list
import pandas as np
import numpy as np
a=[1,2,3,4,5,6]
data=pd.DataFrame(a)
print(data)
```

```python
# creating dataframe from a list of list
import numpy as np
import pandas as pd
a=[1,["a",3],4,["b",5],6]
data=pd.DataFrame(a)
print(data)
```

Write a program that creates a dataframe storing names and marks of students using list of lists. give relevant names to the columns

Write a program that demonstrate deleting columns from a data frame using the del statement and pop method

```python
import numpy as np
import pandas as pd
data={'Kiara':pd.Series([78,89,60],index=['M1','M2','M3']),
    'Meera':pd.Series([76,89,90,95],index=['M1','M2','M3','M4']),
    'Shaina':pd.Series([23,56,78,80],index=['M1','M2','M3','M4']),
    'Herman':pd.Series([45,56,78,80],index=['M1','M2','M3','M4'])}
df=pd.DataFrame(data)
print(df)
print()
del df['Shaina']
print (df)
print()
df.pop('Herman')
print(df)
```
