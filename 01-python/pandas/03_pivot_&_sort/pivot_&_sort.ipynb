### Pivoting data frame
the pivot() function is used to reshape a given DataFrame object organized by index/column values.
DataFrame.pivot(self,index=none,columns=none,values=none)


```python
import pandas as pd
data={
    'name':['Christ','moise','malick','zanga','Mari'],
    'degree':['master','graduate','undergraduate','undergraduate','Master'],
    'age':[30,26,23,24,31]
}
df=pd.DataFrame(data)
print(df)
```

```python
df1=df.pivot(index='name',columns='degree',values='age')
print(df1)
print()
df1=df1.reset_index()
print(df1)
```
