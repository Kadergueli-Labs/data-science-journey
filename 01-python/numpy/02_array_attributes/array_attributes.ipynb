### ARRAY ATTRIBUTES
In NumPy, array attributes are intrinsic properties of the ndarray object that provide information and metadata about its structure, shape, and data type. Unlike methods, attributes do not require parentheses () and are accessed directly using dot notation

### Ndim: displays the dimension of the arrray
### Shape: returns a tuple pf integers indicating the size of the arrayy
### Size: retuns the total number of elements int the numPy array
### Dtype: returns the type of the elements in the array
### itemsize: returns the size in bytes of each item
### Reshape: reshapes the numPy array


```python
import numpy as np
a=np.empty((3,4))
print(a)
```

```python
print(np.ndim(a))
```

```python
print(np.shape(a))
```

```python
print(np.size(a))
```

```python
print(a.dtype)
```
